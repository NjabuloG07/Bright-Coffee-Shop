```sql
SELECT *
FROM workspace.default.`brightcoffee_consultation`;


--Query 1: Viewing first rows of the dataset
select *
from `workspace`.`default`.`brightcoffee_consultation`
limit 10;


--Query 2: Viewing distinct store locations
select distinct store_location
from `workspace`.`default`.`brightcoffee_consultation`;


--Query 3: Selecting specific columns
select distinct transaction_id,
                product_category,
                unit_price
from `workspace`.`default`.`brightcoffee_consultation`
limit 20;


--Query 4: using a column alias
select product_category AS Category,
       unit_price AS Price
from `workspace`.`default`.`brightcoffee_consultation`
limit 100;


--Query 5: Filtering using WHERE
select *
from `workspace`.`default`.`brightcoffee_consultation`
WHERE product_category = 'Coffee';


--Query 6: Filtering using WHERE and AND
select *
from `workspace`.`default`.`brightcoffee_consultation`
WHERE product_category = 'coffee beans' AND unit_price > 10;


-- Query 7: Finding the minimum unit price
select MIN(unit_price) AS Min_Price
from `workspace`.`default`.`brightcoffee_consultation`;


--Query 8: Finding the maximum unit price
select MAX(unit_price) AS Max_Price
from `workspace`.`default`.`brightcoffee_consultation`;


--Query 9: Filtering using WHERE and OR
select *
from `workspace`.`default`.`brightcoffee_consultation`
WHERE product_category = 'Coffee' OR product_category = 'Tea';


--Query 10: Filtering using WHERE and IN
select *
from `workspace`.`default`.`brightcoffee_consultation`
WHERE product_category IN ('Coffee','Tea');


--Query 11: Sorting results
select product_category,
       unit_price
from `workspace`.`default`.`brightcoffee_consultation`
ORDER BY unit_price DESC;



--Query 12: Counting total transactions
select count(*)
from `workspace`.`default`.`brightcoffee_consultation`;



--Query 13: Calculate total sales
select sum(transaction_qty* unit_price) AS Total_Sales
from `workspace`.`default`.`brightcoffee_consultation`;


--Query 14: Calculate minimum and maximum price
select MIN(unit_price) AS cheapest_item,
       MAX(unit_price) AS most_expensive_item
from `workspace`.`default`.`brightcoffee_consultation`;


--Query 15: Total sales per category
select product_category,
       SUM(transaction_qty* unit_price) AS Total_Sales
from `workspace`.`default`.`brightcoffee_consultation`
group by product_category
order by Total_Sales DESC;    


-- Query 16: Adding columns to enhance insights
select 
    transaction_id,
    transaction_date,
    transaction_time,
    transaction_qty,
    store_id,
    store_location,
    product_id,
    unit_price,
    product_category,
    product_type,
    product_detail,

    case 
        when transaction_time < '11:59:59' then 'Morning'
        when transaction_time < '17:59:59' then 'Afternoon'
        else 'Evening'
    end as time_bucket,

    case 
        when Dayname(transaction_date) in ('Saturday', 'Sunday') then 'Weekend'
        else 'Weekday'
    end as day_classification,

    case 
        when (transaction_qty*unit_price) <= 50 then 'Low Spender'
        when (transaction_qty*unit_price) between 51 and 200 then 'Medium Spender'
        when (transaction_qty*unit_price) between 201 and 300 then 'High Spender'
        else 'Super Spender'
    end as Spend_bucket,

transaction_qty* unit_price AS Total_Revenue

from `workspace`.`default`.`brightcoffee_consultation`;







```
